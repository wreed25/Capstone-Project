```python
# Inspect contract history for a sample player

sample_player = "00-0032463"

contracts.filter(
    pl.col("gsis_id") == sample_player
).sort("year_signed")
```

```python
# Inspect duplicate contract records for wide receivers

wr_contracts = contracts.filter(
    pl.col("position") == "WR"
)

duplicate_summary = (
    wr_contracts
    .group_by(["gsis_id", "year_signed"])
    .agg(
        pl.len().alias("contract_rows"),
        pl.col("apy").n_unique().alias("unique_apy_values"),
        pl.col("team").n_unique().alias("unique_teams")
    )
    .filter(
        pl.col("contract_rows") > 1
    )
    .sort(
        "contract_rows",
        descending=True
    )
)

print("Wide receiver contract rows:", wr_contracts.height)
print(
    "Player and signing year combinations with multiple rows:",
    duplicate_summary.height
)

duplicate_summary.head(20)
```

```python
# Quantify usable WR contract records and exact duplicates

wr_contracts_valid = wr_contracts.filter(
    pl.col("gsis_id").is_not_null()
)

contract_core_columns = [
    "gsis_id",
    "year_signed",
    "team",
    "years",
    "value",
    "apy",
    "guaranteed",
    "apy_cap_pct",
    "inflated_value",
    "inflated_apy",
    "inflated_guaranteed",
]

wr_contracts_unique = (
    wr_contracts_valid
    .unique(
        subset=contract_core_columns,
        maintain_order=True
    )
)

print("All WR contract rows:", wr_contracts.height)
print(
    "WR rows with valid gsis_id:",
    wr_contracts_valid.height
)
print(
    "WR rows removed for missing gsis_id:",
    wr_contracts.height - wr_contracts_valid.height
)

print(
    "\nExact duplicate contract rows removed:",
    wr_contracts_valid.height - wr_contracts_unique.height
)

print(
    "Unique usable WR contract rows:",
    wr_contracts_unique.height
)

print(
    "Unique WR players with contract records:",
    wr_contracts_unique["gsis_id"].n_unique()
)
```

```python
# Recheck multiple contracts after removing missing IDs and exact duplicates

wr_contract_year_summary = (
    wr_contracts_unique
    .group_by(["gsis_id", "year_signed"])
    .agg(
        pl.len().alias("contract_rows"),
        pl.col("apy").n_unique().alias("unique_apy_values"),
        pl.col("team").n_unique().alias("unique_teams")
    )
    .filter(
        pl.col("contract_rows") > 1
    )
    .sort(
        "contract_rows",
        descending=True
    )
)

print(
    "Player and signing year combinations with multiple distinct contracts:",
    wr_contract_year_summary.height
)

print(
    "Maximum distinct contract rows for one player and signing year:",
    wr_contract_year_summary["contract_rows"].max()
)

wr_contract_year_summary.head(20)
```

```python
# Inspect one WR with multiple distinct contracts in the same year

sample_wr = "00-0032208"
sample_year = 2021

(
    wr_contracts_unique
    .filter(
        (pl.col("gsis_id") == sample_wr)
        & (pl.col("year_signed") == sample_year)
    )
    .select([
        "player",
        "position",
        "team",
        "year_signed",
        "years",
        "value",
        "apy",
        "guaranteed",
        "apy_cap_pct",
        "inflated_apy",
        "is_active",
    ])
    .sort("apy", descending=True)
)
```

```python
# Check contract ambiguity after including team

wr_contract_team_year_summary = (
    wr_contracts_unique
    .group_by([
        "gsis_id",
        "year_signed",
        "team"
    ])
    .agg(
        pl.len().alias("contract_rows"),
        pl.col("apy").n_unique().alias("unique_apy_values"),
        pl.col("years").n_unique().alias("unique_contract_lengths")
    )
    .filter(
        pl.col("contract_rows") > 1
    )
    .sort(
        "contract_rows",
        descending=True
    )
)

print(
    "Player, team, and signing year combinations with multiple contracts:",
    wr_contract_team_year_summary.height
)

print(
    "Maximum contracts within one player, team, and signing year:",
    wr_contract_team_year_summary["contract_rows"].max()
)

wr_contract_team_year_summary.head(20)
```

```python
# Inspect detailed contract history for one WR

sample_wr_history = (
    wr_contracts_unique
    .filter(
        pl.col("gsis_id") == "00-0032208"
    )
    .select([
        "player",
        "gsis_id",
        "contract_history"
    ])
    .head(1)
)

print(sample_wr_history)

print("\nContract history contents:")
print(
    sample_wr_history[
        "contract_history"
    ][0]
)
```

```python
# Expand WR contract history into one row per contract event

wr_contract_history = (
    wr_contracts_unique
    .select([
        "gsis_id",
        "player",
        "position",
        "contract_history"
    ])
    .explode("contract_history")
    .unnest("contract_history")
)

print("Expanded WR contract history rows:", wr_contract_history.height)
print("\nColumns:")
print(wr_contract_history.columns)

wr_contract_history.head(10)
```

```python
# Expand one contract history per WR into contract events

wr_player_histories = (
    wr_contracts_valid
    .select([
        "gsis_id",
        "player",
        "position",
        "contract_history"
    ])
    .unique(
        subset=["gsis_id"],
        maintain_order=True
    )
)

wr_contract_history = (
    wr_player_histories
    .explode("contract_history")
    .unnest("contract_history")
    .unique(
        subset=[
            "gsis_id",
            "team",
            "contract_type",
            "status",
            "year_signed",
            "yrs",
            "total",
            "apy",
            "guarantees",
            "amount_earned",
            "percent_earned",
            "effective_apy"
        ],
        maintain_order=True
    )
)

print(
    "Unique WR players:",
    wr_player_histories.height
)

print(
    "Unique WR contract events:",
    wr_contract_history.height
)

wr_contract_history.head(10)
```

```python
# Review WR contract event types and statuses

contract_type_counts = (
    wr_contract_history
    .group_by("contract_type")
    .agg(
        pl.len().alias("events")
    )
    .sort(
        "events",
        descending=True
    )
)

status_counts = (
    wr_contract_history
    .group_by("status")
    .agg(
        pl.len().alias("events")
    )
    .sort(
        "events",
        descending=True
    )
)

print("Contract types:")
print(contract_type_counts)

print("\nContract statuses:")
print(status_counts)
```

```python
status_counts.to_dicts()
```

```python
# Review contract history coverage by signing year

contract_year_coverage = (
    wr_contract_history
    .group_by("year_signed")
    .agg(
        pl.len().alias("contract_events"),
        pl.col("gsis_id").n_unique().alias("unique_players")
    )
    .sort("year_signed")
)

print(contract_year_coverage)
```

```python
# Focus contract coverage on the project analysis period

project_contract_coverage = (
    wr_contract_history
    .filter(
        pl.col("year_signed").is_between(2012, 2024)
    )
    .group_by("year_signed")
    .agg(
        pl.len().alias("contract_events"),
        pl.col("gsis_id").n_unique().alias("unique_players"),
        pl.col("apy").median().alias("median_apy"),
        pl.col("apy").max().alias("max_apy")
    )
    .sort("year_signed")
)

print(project_contract_coverage)
```
