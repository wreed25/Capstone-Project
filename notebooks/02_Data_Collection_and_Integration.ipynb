```python
nfl_master = (
    player_stats
    .join(
        players,
        left_on="player_id",
        right_on="gsis_id",
        how="left",
        suffix="_player"
    )
    .join(
        contracts,
        left_on="player_id",
        right_on="gsis_id",
        how="left",
        suffix="_contract"
    )
)
```

## 7. Contract Duplication

The merged dataset contains more rows than the player statistics dataset. The increase is caused by multiple contract records for individual players. 

```python
contract_counts = (
    contracts
    .group_by("gsis_id")
    .len()
    .sort("len", descending=True)
)

contract_counts.head(20)
```

```python
# Count missing GSIS IDs in the contracts table

null_gsis = contracts.filter(pl.col("gsis_id").is_null())

print(f"Contract records: {contracts.height:,}")
print(f"Missing gsis_id: {null_gsis.height:,}")
print(f"Percent missing: {100 * null_gsis.height / contracts.height:.2f}%")
```

```python
# Remove records without a GSIS ID and examine duplicates again

contracts_valid = contracts.filter(pl.col("gsis_id").is_not_null())

contract_counts = (
    contracts_valid
    .group_by("gsis_id")
    .len()
    .sort("len", descending=True)
)

contract_counts.head(20)
```

```python
# Count contract records with missing gsis ID's

null_gsis = contracts.filter(
    pl.col("gsis_id").is_null()
)

print(f"Total contract records: {contracts.height:,}")
print(f"missing gsis_id: {null_gsis.height:,}")
print(
    f"Percent missing: "f"{100 * null_gsis.height / contracts.height:.2f}%"
)
```

```python
# Retain contract records with a valid GSIS ID

contracts_valid = contracts.filter(
    pl.col("gsis_id").is_not_null()
)

# Count contract records for each identified player

contract_counts_valid = (
    contracts_valid
    .group_by("gsis_id")
    .len(name="contract_record_count")
    .sort("contract_record_count", descending=True)
)

contract_counts_valid.head(20)
```

### Findings Summary

The contracts dataset was examined to understand the increase in rows following integration with the player statistics dataset.

findings:

- 4,190 contract records (8.10%) contain missing `gsis_id` values and cannot be linked to player statistics.
- Many players legitimately possess multiple contract records because they signed multiple contracts throughout their careers.
- The contracts dataset also contains exact duplicate records. A comparison of individual player histories confirmed that several contracts are duplicated without any differences across the available fields.
- The increased row count after merging is attributable to both, one-to-many relationships and duplicate records present in the source contracts dataset.

No records were modified or removed yet. Data cleaning decisions will be deferred revisited during preprocessing stage after EDA.

## Summary

In this notebook, the data collection and integration phase established the foundation for the predictive modeling workflow. Multiple datasets containing player statistics, player metadata, and contract information were loaded, inspected, and evaluated for quality and compatibility.

Key validation steps confirmed the structure of the player statistics data, verified unique player season observations, and examined identifier consistency across datasets. Rather than introducing duplicated records into the analytical dataset, the contract merge was deferred until a season aware integration strategy can be developed.

Validation confirmed that the resulting dataset contains one observation per player per season with no duplicate player-season combinations.

The completed dataset provides a reliable foundation for exploratory data analysis (EDA), feature engineering, and predictive modeling in the subsequent notebooks.
